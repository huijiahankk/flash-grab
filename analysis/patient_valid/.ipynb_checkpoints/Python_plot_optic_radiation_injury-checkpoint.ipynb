```python
print('This is only for ploting optic injury patient data')

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# Load the data from the Excel file
file_path = '/Users/huijiajia/Documents/matlab/flash-grab/analysis/patient_valid/optic_injury_patient_data.xlsx' # optic_injury_patient_data.xlsx
patient_data = pd.read_excel(file_path)

# Recalculate the column averages without the first column (which is empty)
column_averages = patient_data.iloc[:, 1:].mean()

# Define the columns for 'Grab-in', 'Perceived-in', 'Grab-out', and 'Perceived-out'
grab_in_col = 'Physical location(Grab-in)'
perceived_in_col = 'Perceived location(Grab-in)'
grab_out_col = 'Physical location(Grab-out)'
perceived_out_col = 'Perceived location(Grab-out)'

# Get the indices of the specified columns
grab_in_idx = patient_data.columns.get_loc(grab_in_col) - 1
perceived_in_idx = patient_data.columns.get_loc(perceived_in_col) - 1
grab_out_idx = patient_data.columns.get_loc(grab_out_col) - 1
perceived_out_idx = patient_data.columns.get_loc(perceived_out_col) - 1

# Print the indices to ensure they are correct
print(f"Indices - Grab-in: {grab_in_idx}, Perceived-in: {perceived_in_idx}, Grab-out: {grab_out_idx}, Perceived-out: {perceived_out_idx}")

# Define the adjusted positions for the bars
adjusted_positions = list(range(len(patient_data.columns[1:])))
adjusted_positions[grab_in_idx] = perceived_in_idx - 0.5
adjusted_positions[perceived_in_idx] = perceived_in_idx
adjusted_positions[grab_out_idx] = perceived_out_idx - 0.5
adjusted_positions[perceived_out_idx] = perceived_out_idx

# Sort the positions and corresponding labels
sorted_positions_labels = sorted(zip(adjusted_positions, patient_data.columns[1:]), key=lambda x: x[0])
sorted_positions = [pos for pos, label in sorted_positions_labels]
sorted_labels = [label for pos, label in sorted_positions_labels]

# Define colors for the bars
colors = ['lightblue' if i in [perceived_in_idx, perceived_out_idx] else 'salmon' for i in range(len(patient_data.columns[1:]))]
edge_colors = ['lightblue' if i in [perceived_in_idx, perceived_out_idx] else 'salmon' for i in range(len(patient_data.columns[1:]))]

# Print the colors list to ensure it's correct
print(f"Colors: {colors}")
print(f"Edge Colors: {edge_colors}")

for label, mean_value in zip(sorted_labels, column_averages):
    print(f"Mean value of {label}: {mean_value}")

# Set up the plot
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot the average data of each column with adjusted positions and specified colors
bars = ax1.bar(sorted_positions, column_averages, color=colors, edgecolor=edge_colors, linewidth=1, width=0.5, alpha=0.7)

# Annotate each bar with the mean value
# for bar, mean_value in zip(bars, column_averages):
#     height = bar.get_height()
#     ax1.annotate(f'{mean_value:.2f}',
#                  xy=(bar.get_x() + bar.get_width() / 2, height),
#                  xytext=(0, 3),  # 3 points vertical offset
#                  textcoords="offset points",
#                  ha='center', va='bottom')


# Set the y-axis limits for the primary y-axis
ax1.set_ylim(-45, 70)

# Set the left y-axis label and tick parameters
ax1.set_ylabel('Physical location of the bar (from the blind field)', fontsize=12, color='black')
ax1.tick_params(axis='y', labelcolor='salmon')

# Create a secondary y-axis
ax2 = ax1.twinx()

# Set the y-axis limits for the secondary y-axis
ax2.set_ylim(-45, 70)

# Set the right y-axis label and tick parameters
ax2.set_ylabel('Perceived location of the bar (from the blind field)', fontsize=12, color='black')
ax2.tick_params(axis='y', labelcolor='lightblue')

# Plot each value with a colored dot on the secondary y-axis
for i, row in patient_data.iterrows():
    # Adjust the x-coordinates slightly to prevent overlap (reduce the spread)
    x_coords = [pos + (0.02 * (i % 10 - 5)) for pos in sorted_positions]
    ax2.scatter(x_coords, row[1:], color='grey', alpha=0.7)

# Customize the x-axis labels to match the adjusted column names, excluding the first column
ax1.set_xticks(sorted_positions)
ax1.set_xticklabels(sorted_labels, rotation=45, fontsize=12)

# Adjust layout and remove title
plt.tight_layout()
plt.title('', fontsize=20)

# Remove the top and right spines (borders) of the primary y-axis
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

plt.savefig('/Users/huijiajia/Documents/matlab/flash-grab/analysis/patient_valid/optic_radiation_injury_from_python.pdf', format='pdf')

# Show the plot
plt.show()
```
