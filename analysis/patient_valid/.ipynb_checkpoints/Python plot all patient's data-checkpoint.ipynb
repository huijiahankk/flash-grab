```python
# combined control condition 
print('This is only for plotting all the cortical blindness patients data')

import matplotlib  
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np

# Load the data from the Excel file
file_path = '/Users/huijiajia/Documents/neuroscience/psychological_and_brain_sciences/cortialBlindness/figures/data_analysis_python_code/all_patient_combined_control.xlsx'
patient_data = pd.read_excel(file_path)

# Recalculate the column averages without the first column (which may contain non-numeric data)
column_averages = patient_data.iloc[:, 1:].mean()

# Identify the indices of the specified columns
grab_in_col = 'Physical location(Grab-in)'
perceived_in_col = 'Perceived location(Grab-in)'
grab_out_col = 'Physical location(Grab-out)'
perceived_out_col = 'Perceived location(Grab-out)'

grab_in_idx = patient_data.columns.get_loc(grab_in_col) - 1
perceived_in_idx = patient_data.columns.get_loc(perceived_in_col) - 1
grab_out_idx = patient_data.columns.get_loc(grab_out_col) - 1
perceived_out_idx = patient_data.columns.get_loc(perceived_out_col) - 1

# Print the indices to ensure they are correct
print(f"Indices - Grab-in: {grab_in_idx}, Perceived-in: {perceived_in_idx}, Grab-out: {grab_out_idx}, Perceived-out: {perceived_out_idx}")

# Define the adjusted positions for the bars
adjusted_positions = list(range(len(patient_data.columns[1:])))
adjusted_positions[grab_in_idx] = perceived_in_idx - 0.5
adjusted_positions[perceived_in_idx] = perceived_in_idx
adjusted_positions[grab_out_idx] = perceived_out_idx - 0.5
adjusted_positions[perceived_out_idx] = perceived_out_idx

# Sort the positions and corresponding labels
sorted_positions_labels = sorted(zip(adjusted_positions, patient_data.columns[1:]), key=lambda x: x[0])
sorted_positions = [pos for pos, label in sorted_positions_labels]
sorted_labels = [label for pos, label in sorted_positions_labels]

# Debug: Print the sorted labels to check them
print(f"Sorted Labels: {sorted_labels}")

# Create a consistent color map for the prefixes for scatter points
prefixes = set(row.iloc[0].rsplit('-', 1)[0] for _, row in patient_data.iterrows())
colormap = matplotlib.colormaps['plasma']
prefix_color_map = {prefix: colormap(i / (len(prefixes) - 1)) for i, prefix in enumerate(prefixes)}

colors = ['lightblue' if i in [perceived_in_idx, perceived_out_idx] else 'salmon' for i in range(len(patient_data.columns[1:]))]
edge_colors = ['lightblue' if i in [perceived_in_idx, perceived_out_idx] else 'salmon' for i in range(len(patient_data.columns[1:]))]


# # Add a default color for any unexpected prefixes
# default_color = 'grey'

# # Assign colors based on the label for bars
# def get_bar_color(label):
#     if 'Control' in label and ('Grab-in' in label or 'Grab-out' in label):
#         return 'salmon'
#     elif 'Perceived location(Grab-in)' in label or 'Perceived location(Grab-out)' in label:
#         return 'lightblue'
#     return 'lightblue'  # Default to lightblue for any other case

default_color = 'salmon'
# Assign colors for scatter based on the prefix
def get_scatter_color(prefix):
    return prefix_color_map.get(prefix, default_color)

# # Debug: Print the labels to check if they match correctly
# for label in patient_data.columns[1:]:
#     print(f"Label: {label}, Bar Color: {get_bar_color(label)}, Scatter Color: {get_scatter_color(label.split('-')[0])}")

# bar_colors = [get_bar_color(label) for label in patient_data.columns[1:]]

# Set up the plot
plt.figure(figsize=(12, 6))

# Plot the average data of each column with adjusted positions and specified colors
bars = plt.bar(sorted_positions, column_averages, color=colors, edgecolor=edge_colors,linewidth=1, width=0.5, alpha=0.7)

# Plot each value with a colored dot based on its prefix
scatter_handles = []
scatter_labels = []
for i, row in patient_data.iterrows():
    prefix = row.iloc[0].rsplit('-', 1)[0]
    full_label = row.iloc[0]
    color = get_scatter_color(prefix)
    # Add a small random horizontal shift to each scatter point
    x_coords = [pos + np.random.uniform(-0.05, 0.05) for pos in sorted_positions]
    scatter = plt.scatter(x_coords, row.iloc[1:], color=color, alpha=0.7, label=full_label)
    if full_label not in scatter_labels:
        scatter_handles.append(scatter)
        scatter_labels.append(full_label)

# Customize the y-axis label
plt.ylabel('Bar shift degree from blind field', fontsize=14)

# Set x-axis labels to match the sorted column names
plt.xticks(ticks=sorted_positions, labels=sorted_labels, rotation=45, fontsize=14)

# Create custom legend for unique labels
plt.legend(scatter_handles, scatter_labels, bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=14) # title="Labels"

# Adjust layout and remove title
plt.tight_layout()
plt.title('', fontsize=20)

# Remove the top and right spines (borders)
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.set_ylim(-30, 65)

plt.savefig('/Users/huijiajia/Documents/neuroscience/psychological_and_brain_sciences/cortialBlindness/figures/data_analysis_python_code/all_patient_figure_from_python.pdf', format='pdf')

# Show the plot
plt.show()

# Calculate the column averages without the first column
column_averages = patient_data.iloc[:, 1:].mean()

# Print the average value of each column
for column, average in column_averages.items():
    print(f"Average value in {column}: {average}")


```

```python
# seperated control condition 
print('This is only for plotting all the cortical blindness patients data')

import matplotlib  
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np

# Load the data from the Excel file
file_path = '/Users/huijiajia/Documents/neuroscience/psychological_and_brain_sciences/cortialBlindness/figures/data_analysis_python_code/all_patient_seperated_control.xlsx'
patient_data = pd.read_excel(file_path)

# Recalculate the column averages without the first column (which may contain non-numeric data)
column_averages = patient_data.iloc[:, 1:].mean()

# Get the indices of the specified columns
control_in_idx = patient_data.columns.get_loc('Control(Grab-in)')
grab_in_idx = patient_data.columns.get_loc('Physical location(Grab-in)')
perceived_in_idx = patient_data.columns.get_loc('Perceived location(Grab-in)')
control_out_idx = patient_data.columns.get_loc('Control(Grab-out)')
grab_out_idx = patient_data.columns.get_loc('Physical location(Grab-out)')
perceived_out_idx = patient_data.columns.get_loc('Perceived location(Grab-out)')


# Initialize the list to have enough elements to accommodate all columns excluding the first one
num_columns = len(patient_data.columns) - 1
adjusted_positions = [None] * num_columns

# Assign adjusted positions
adjusted_positions[control_in_idx - 1] = control_in_idx + 0.5  # Adjust for the excluded first column
adjusted_positions[grab_in_idx - 1] = grab_in_idx  
adjusted_positions[perceived_in_idx - 1] = perceived_in_idx - 0.5
adjusted_positions[control_out_idx - 1] = control_out_idx + 0.5 
adjusted_positions[grab_out_idx - 1] = grab_out_idx 
adjusted_positions[perceived_out_idx - 1] = perceived_out_idx - 0.5


# Sort the positions and corresponding labels
sorted_positions_labels = sorted(zip(adjusted_positions, patient_data.columns[1:]), key=lambda x: x[0])
sorted_positions = [pos for pos, label in sorted_positions_labels]
sorted_labels = [label for pos, label in sorted_positions_labels]

# Debug: Print the sorted labels to check them
print(f"Sorted Labels: {sorted_labels}")

# Create a consistent color map for the prefixes for scatter points
prefixes = set(row.iloc[0].rsplit('-', 1)[0] for _, row in patient_data.iterrows())
colormap = matplotlib.colormaps['plasma']
prefix_color_map = {prefix: colormap(i / (len(prefixes) - 1)) for i, prefix in enumerate(prefixes)}

colors = ['lightblue' if i in [perceived_in_idx-1, perceived_out_idx-1] else 'salmon' for i in range(len(patient_data.columns[1:]))]
edge_colors = ['lightblue' if i in [perceived_in_idx-1, perceived_out_idx-1] else 'salmon' for i in range(len(patient_data.columns[1:]))]



default_color = 'salmon'
# Assign colors for scatter based on the prefix
def get_scatter_color(prefix):
    return prefix_color_map.get(prefix, default_color)


# Set up the plot
plt.figure(figsize=(12, 6))

# Plot the average data of each column with adjusted positions and specified colors
bars = plt.bar(sorted_positions, column_averages, color=colors, edgecolor=edge_colors,linewidth=1, width=0.5, alpha=0.7)

# Plot each value with a colored dot based on its prefix
scatter_handles = []
scatter_labels = []
for i, row in patient_data.iterrows():
    prefix = row.iloc[0].rsplit('-', 1)[0]
    full_label = row.iloc[0]
    color = get_scatter_color(prefix)
    # Add a small random horizontal shift to each scatter point
    x_coords = [pos + np.random.uniform(-0.05, 0.05) for pos in sorted_positions]
    scatter = plt.scatter(x_coords, row.iloc[1:], color=color, alpha=0.7, label=full_label)
    if full_label not in scatter_labels:
        scatter_handles.append(scatter)
        scatter_labels.append(full_label)

# Customize the y-axis label
plt.ylabel('Bar shift from blind field', fontsize=14)

# Set x-axis labels to match the sorted column names
plt.xticks(ticks=sorted_positions, labels=sorted_labels, rotation=45, fontsize=14)

# Create custom legend for unique labels
plt.legend(scatter_handles, scatter_labels, bbox_to_anchor=(1.05, 1), loc='upper left', fontsize=14) # title="Labels"

# Adjust layout and remove title
plt.tight_layout()
plt.title('', fontsize=20)

# Remove the top and right spines (borders)
ax = plt.gca()
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

ax.set_ylim(-30, 65)

plt.savefig('/Users/huijiajia/Documents/neuroscience/psychological_and_brain_sciences/cortialBlindness/figures/data_analysis_python_code/all_patient_figure_from_python.pdf', format='pdf')

# Show the plot
plt.show()

# Calculate the column averages without the first column
column_averages = patient_data.iloc[:, 1:].mean()

# Print the average value of each column
for column, average in column_averages.items():
    print(f"Average value in {column}: {average}")


```
