```python
import matplotlib  
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm

# Load the data from the Excel file
file_path = '/Users/huijiajia/Documents/matlab/flash-grab/analysis/patient_valid/all_patient_data.xlsx'
patient_data = pd.read_excel(file_path)

# Recalculate the column averages without the first column (which may contain non-numeric data)
column_averages = patient_data.iloc[:, 1:].mean()

# Set up the plot
plt.figure(figsize=(12, 6))

# Plot the average data of each column with unfilled bars (transparent fill, custom light blue edges)
bars = plt.bar(patient_data.columns[1:], column_averages, color="red", edgecolor="red", linewidth=1,width=0.5,alpha=0.3)  # width=0.5


# Overlay unfilled blue rectangles
for bar in bars:
    plt.gca().add_patch(plt.Rectangle((bar.get_x(), bar.get_y()), bar.get_width(), bar.get_height(), fill=False, edgecolor="red", linewidth=2))

# Extract unique prefixes from the first column
prefixes = set(row[0].split('-')[0] for _, row in patient_data.iterrows())

# Create a colormap and map each prefix to a color
# This replaces the deprecated use of cm.get_cmap with the newer approach
colormap = matplotlib.colormaps['viridis']  # Access the colormap using the new method
prefix_color_map = {prefix: colormap(i / (len(prefixes) - 1)) for i, prefix in enumerate(prefixes)}  # Normalize color mapping


# Plot each value with a colored dot based on its prefix
for i, row in patient_data.iterrows():
    prefix = row[0].split('-')[0]
    color = prefix_color_map[prefix]
    plt.scatter(patient_data.columns[1:], row[1:], color=color, label=row[0])

# Customize the y-axis label
plt.ylabel(' --->> Away from blind field',fontsize=18)

# Set x-axis labels to match the column names, excluding the first column
plt.xticks(ticks=range(len(patient_data.columns[1:])), labels=patient_data.columns[1:], rotation=45,fontsize=20)

# Handle legend: create custom labels for unique prefixes
handles, labels = plt.gca().get_legend_handles_labels()
unique_labels = list(dict.fromkeys(labels))
unique_handles = [handles[labels.index(label)] for label in unique_labels]
plt.legend(unique_handles, unique_labels, bbox_to_anchor=(1.05, 1), loc='upper left',fontsize=20)   # title="Patient Condition"

# Adjust layout and remove title
plt.tight_layout()
plt.title('',fontsize=16)

# Show the plot
plt.show()


# Calculate the column averages without the first column
column_averages = patient_data.iloc[:, 1:].mean()

# Print the average value of each column
for column, average in column_averages.items():
    print(f"Average value in {column}: {average}")


```

```python
import matplotlib  
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np



# Load the data
file_path = '/Users/huijiajia/Documents/matlab/flash-grab/analysis/patient_valid/all_patient_data.xlsx'
patient_data = pd.read_excel(file_path)

# Recalculate the column averages
column_averages = patient_data.iloc[:, 1:].mean()

# Placeholder for unique legend labels
unique_labels = {}

# Set up the plot
plt.figure(figsize=(12, 6))
bars = plt.bar(patient_data.columns[1:], column_averages, color="red", edgecolor="red", linewidth=1,width=0.5)

# Define different shapes and whether they are filled
prefix_shapes = {
    'ml': 'o',  # Circle for both 'ml-upper' and 'ml-lower'
    'mgq': '^',  # Triangle for 'mgq-upper' and 'mgq-lower'
    'yfq': 's',  # Square for 'yfq-upper' and 'yfq-lower'
    'zhuerl': 'D',  # Diamond for 'zhuerl-upper' and 'zhuerl-lower'
    'll': 'P',  # Plus (filled) for 'll-upper' and 'll-lower'
    # Add more prefixes as necessary
}

fill_styles = {
    'upper': 'full',
    'lower': 'none'
}

column_sems = patient_data.iloc[:, 1:].sem()  # Standard Error of the Mean for each column

# Plotting
for i, row in patient_data.iterrows():
    prefix, condition = row[0].split('-')
    shape = prefix_shapes.get(prefix, 'o')  # Default to 'o' if prefix not found
    fill = fill_styles.get(condition, 'none')  # Default to 'none' if condition not found
    
    # Set marker style based on condition
    marker_style = shape  # Shape determined by prefix
    fill_style = 'full' if fill == 'full' else 'none'  # Fill style determined by condition
    
    plt.scatter(patient_data.columns[1:], row[1:], marker=marker_style, 
                facecolors='none' if fill_style == 'none' else 'blue', 
                edgecolors='black', s=100, label=row[0])  # Adjust size as needed

    # Recalculate the column averages and SEMs
column_averages = patient_data.iloc[:, 1:].mean()
column_sems = patient_data.iloc[:, 1:].sem()  # Standard Error of the Mean for each column


# Plot the averages with error bars on the same figure
plt.errorbar(patient_data.columns[1:], column_averages, yerr=column_sems, fmt='o', ecolor="red", capsize=5, color="red", markersize=8)

# Customize the plot
plt.ylabel('Score')
plt.xticks(ticks=range(len(patient_data.columns[1:])), labels=patient_data.columns[1:], rotation=45)
plt.title('Flash grab along the blind field border of hemianopia patients')
plt.tight_layout()  # Adjust the layout to not overlap elements
plt.show()


```
