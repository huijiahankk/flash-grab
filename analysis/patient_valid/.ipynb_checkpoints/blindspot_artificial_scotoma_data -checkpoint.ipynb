```python
# seperated control condition

import matplotlib
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np

# Load the data from the Excel file
file_path = '/Users/huijiajia/Documents/matlab/flash-grab/analysis/eyeGuiding_behavior_data/blind_spot_data_seperated_control.xlsx'
patient_data = pd.read_excel(file_path)

# Recalculate the column averages without the first column (which may contain non-numeric data)
column_averages = patient_data.iloc[:, 1:].mean()

# Get the indices of the specified columns
control_in_idx = patient_data.columns.get_loc('Control(Grab-in)')
grab_in_idx = patient_data.columns.get_loc('Physical location(Grab-in)')
perceived_in_idx = patient_data.columns.get_loc('Perceived location(Grab-in)')
control_out_idx = patient_data.columns.get_loc('Control(Grab-out)')
grab_out_idx = patient_data.columns.get_loc('Physical location(Grab-out)')
perceived_out_idx = patient_data.columns.get_loc('Perceived location(Grab-out)')


# Initialize the list to have enough elements to accommodate all columns excluding the first one
num_columns = len(patient_data.columns) - 1
adjusted_positions = [None] * num_columns

# Assign adjusted positions
adjusted_positions[control_in_idx - 1] = control_in_idx + 0.5  # Adjust for the excluded first column
adjusted_positions[grab_in_idx - 1] = grab_in_idx  
adjusted_positions[perceived_in_idx - 1] = perceived_in_idx - 0.5
adjusted_positions[control_out_idx - 1] = control_out_idx + 0.5 
adjusted_positions[grab_out_idx - 1] = grab_out_idx 
adjusted_positions[perceived_out_idx - 1] = perceived_out_idx - 0.5

# # Fill remaining None values with their respective indices
# adjusted_positions = [i if i is not None else idx for idx, i in enumerate(adjusted_positions)]

# Print adjusted positions for verification
print(f"Adjusted Positions: {adjusted_positions}")

# Sort the positions and corresponding labels
sorted_positions_labels = sorted(zip(adjusted_positions, patient_data.columns[1:]), key=lambda x: x[0])
sorted_positions = [pos for pos, label in sorted_positions_labels]
sorted_labels = [label for pos, label in sorted_positions_labels]

# Define colors for the bars
colors = ['lightblue' if i in [perceived_in_idx - 1, perceived_out_idx - 1] else 'salmon' for i in range(len(patient_data.columns[1:]))]
edge_colors = ['lightblue' if i in [perceived_in_idx - 1, perceived_out_idx - 1] else 'salmon' for i in range(len(patient_data.columns[1:]))]

# Print the colors list to ensure it's correct
print(f"Colors: {colors}")
print(f"Edge Colors: {edge_colors}")

# Set up the plot
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot the average data of each column with adjusted positions and specified colors
bars = ax1.bar(sorted_positions, column_averages, color=colors, edgecolor=edge_colors, linewidth=1, width=0.5, alpha=0.7)

# Set the y-axis limits for the primary y-axis
ax1.set_ylim(-30, 65)

# Set the left y-axis label and tick parameters
ax1.set_ylabel('Bar distance from the blind spot', fontsize=12, color='black')
ax1.tick_params(axis='y', labelcolor='salmon')

# Create a secondary y-axis
ax2 = ax1.twinx()

# Set the y-axis limits for the secondary y-axis
ax2.set_ylim(-30, 65)

# Set the right y-axis label and tick parameters
# ax2.set_ylabel('Bar distance from the artificial scotoma)', fontsize=12, color='black')
ax2.tick_params(axis='y', labelcolor='lightblue')

# Extract unique prefixes from the first column
prefixes = set(row.iloc[0].split('-')[0] for _, row in patient_data.iterrows())

# Create a colormap and map each prefix to a color
colormap = cm.plasma
prefix_color_map = {prefix: colormap(i / (len(prefixes) - 1)) for i, prefix in enumerate(prefixes)}

# Plot each value with a colored dot based on its prefix on the secondary y-axis
for i, row in patient_data.iterrows():
    prefix = row.iloc[0].split('-')[0]
    color = prefix_color_map[prefix]
    x_coords = [pos + np.random.uniform(-0.05, 0.05) for pos in sorted_positions]
    ax2.scatter(x_coords, row.iloc[1:], color=color, alpha=0.7, label=row.iloc[0])

    # ax2.scatter(sorted_positions, row.iloc[1:], color=color, label=row.iloc[0])

# Customize the x-axis labels to match the adjusted column names, excluding the first column
ax1.set_xticks(sorted_positions)
ax1.set_xticklabels(sorted_labels, rotation=45, fontsize=12)

# Handle legend: create custom labels for unique prefixes
handles, labels = ax2.get_legend_handles_labels()
unique_labels = list(dict.fromkeys(labels))
unique_handles = [handles[labels.index(label)] for label in unique_labels]
ax2.legend(unique_handles, unique_labels, bbox_to_anchor=(1.1, 1), loc='upper left', fontsize=12)  # Adjust bbox_to_anchor to move the legend outside

# Adjust layout and remove title
plt.tight_layout()
plt.title('', fontsize=20)

# Remove the top and right spines (borders) of the primary y-axis
ax1.spines['top'].set_visible(False)
ax1.spines['right'].set_visible(False)

# Remove the top spine of the secondary y-axis
ax2.spines['top'].set_visible(False)

# Adjust layout and remove title
plt.tight_layout()
plt.title('', fontsize=20)

# plt.savefig('/Users/huijiajia/Documents/matlab/flash-grab/analysis/eyeGuiding_behavior_data/blind_spot_figure_from_python.pdf', format='pdf')

# Show the plot
plt.show()
```
